```python
import argparse as ap
import datetime as dt
import glob
import numpy as np

from sotodlib import core
from sotodlib.core.metadata.loader import LoaderError
```

```python
start = dt.datetime(2026, 6, 1, tzinfo=dt.timezone.utc)
end = dt.datetime(2026, 6, 27, tzinfo=dt.timezone.utc)
ctx_path = "../../ctxs/"
ctx = core.Context(ctx_path + "smurf_det_preproc.yaml")
obs_list = ctx.obsdb.query(
    f"{end.timestamp()} > timestamp and timestamp > {start.timestamp()} and type=='obs' and subtype=='cmb'"
)
```

```python
new_ctx_path = "../../ctxs/preprocess_260604.yaml"
for i, obs in enumerate(obs_list):
    obs_id = obs["obs_id"]
    cur_ot = str(obs_id).split("_")[2][3:]
    if cur_ot != "o2": continue
    ctx = core.Context(new_ctx_path)
    try:  # Much faster than ctx.get_meta
        det_info = ctx.get_det_info(obs_id)
    except LoaderError:
        print(f"No meta data for obs {obs_id}")
        continue
    wafers = np.unique(det_info["stream_id"])
    if "ufm_mv73" in wafers:
        print("MV73!")
```
